```python
import pandas as pd 
from sqlalchemy import create_engine
```

```python
conn_string = "mysql+pymysql://root:YOUR_PASSWORD@localhost:3306/uber_projects"
engine = create_engine(conn_string)

```

```python
df = pd.read_sql("SELECT * FROM uber_rides_india", engine)
print("Rows imported from MySQL:", len(df))
```

```python
completed = df[df["status"] == "Completed"]
```

```python
payment_counts = completed["payment_method"]
print(payment_counts)
```

```python
print("Q1. Number of rides by payment method")
payment_counts = completed["payment_method"].value_counts()
print(payment_counts)
print()
```

```python
df.dtypes
```

```python
completed["fare_amount"] = pd.to_numeric(completed["fare_amount"], errors='coerce')
print("Q2. Average fare by ride type")
avg_fare = completed.groupby("ride_type")["fare_amount"].mean()
avg_fare = avg_fare.round(2).sort_values(ascending=False)
print(avg_fare)


```

```python

print("Q3. Average fare: normal vs surge")
completed["surge_multiplier"] = pd.to_numeric(completed["surge_multiplier"],errors='coerce')
normal_rides = completed[completed["surge_multiplier"] == 1.0]
surge_rides  = completed[completed["surge_multiplier"] > 1.0]
print("Average fare without surge:", round(normal_rides["fare_amount"].mean(), 2))
print("Average fare with surge:   ", round(surge_rides["fare_amount"].mean(), 2))
print("Number of surge rides:     ", len(surge_rides))
```

```python
print("Q4. Rides on weekdays vs weekends")
weekend_days = ["Saturday", "Sunday"]
df["day_type"] = "Weekday"
df.loc[df["day_of_week"].isin(weekend_days), "day_type"] = "Weekend"
day_type_counts = df["day_type"].value_counts()
print(day_type_counts)
```

```python
print("Q5. Top 10 drivers by number of completed rides")
top_drivers = completed["driver_id"].value_counts().head(10)
print(top_drivers)
print()

print("Analysis complete.")
```
